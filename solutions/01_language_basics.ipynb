## Exercise 2: Indexing

```python
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun",
          "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"]

print(months[0])
print(months[-1])
print(months[:3])
print(months[5:8])
print(months[::3])
print(months[::-1])
```

June is the sixth month, so it sits at position 5. The slice `5:8` takes positions 5, 6 and 7.

## Exercise 3: Dicts

```python
bill_length = {"Adelie": 38.8, "Chinstrap": 48.8}

bill_length["Gentoo"] = 47.5

for name, mm in bill_length.items():
    print(f"{name}: {mm} mm")

longest_name = None
longest_mm = 0
for name, mm in bill_length.items():
    if mm > longest_mm:
        longest_name = name
        longest_mm = mm

print(longest_name)
```

A shorter way to do part 3, once you are comfortable: `max(bill_length, key=bill_length.get)`.

## Exercise 4: Functions

```python
masses = [3750, 3800, 3250, 3450]


def summarize(values):
    return {
        "n": len(values),
        "mean": sum(values) / len(values),
        "min": min(values),
        "max": max(values),
    }


summarize(masses)
```

## Exercise 5: List comprehensions

```python
temps_f = [14, 32, 50, 68, 86]

temps_c = [(f - 32) * 5 / 9 for f in temps_f]
above_freezing = [c for c in temps_c if c > 0]

print(temps_c)
print(above_freezing)
```

## Exercise 6: NumPy

```python
import numpy as np

temps_f = np.array([14, 32, 50, 68, 86])

temps_c = (temps_f - 32) * 5 / 9

print(temps_c)
print(temps_c[temps_c > 0])
print(np.where(temps_c > 0, "above", "not above"))
print(temps_c[(temps_c > 0) & (temps_c < 25)])
```

## Exercise 7: Translate from R

```python
import numpy as np

scores = np.array([88, 92, 79, 65, 95])
curved = scores + 5
print(curved[curved >= 90].mean())
```

`scores` has to be a NumPy array. With a plain list, `scores + 5` is an error.
