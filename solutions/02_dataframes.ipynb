## Exercise 1: Read and inspect

```python
gapminder = pd.read_csv("../data/gapminder.csv")

print(gapminder.shape)
print(gapminder.dtypes)
gapminder.head()
```

## Exercise 2: Series

```python
print(gapminder["lifeExp"].mean())
print(gapminder["gdpPercap"].max())
print(gapminder["country"].nunique())
gapminder["continent"].value_counts()
```

## Exercise 3: Brackets

```python
print(type(gapminder["country"]))
print(type(gapminder[["country"]]))
print(type(gapminder[["country", "year"]]))
```

A single name gives a Series. A list of names gives a DataFrame, even when the list has only one name in it.

## Exercise 4: Add a column

```python
gapminder["gdp_billions"] = gapminder["pop"] * gapminder["gdpPercap"] / 1e9
gapminder.head(3)
```

## Exercise 5: Position and label

```python
print(gapminder.iloc[10:15][["country", "year", "lifeExp"]])
print(gapminder.loc[10:14, ["country", "year", "lifeExp"]])
```

`.iloc[10:15]` excludes position 15, while `.loc[10:14]` includes label 14. Both give five rows.

## Exercise 6: The index after sorting

```python
top = gapminder.sort_values("lifeExp", ascending=False).head(5)

print(top.index)
print(top.iloc[0]["country"])

top.reset_index(drop=True)
```

The index holds each row's label from the original `gapminder`, which was its position there (Japan in 2007 is row 803).

`top.loc[0]` asks for the row *labelled* 0. That row is Afghanistan in 1952, which is not in `top`, so it raises a `KeyError`. `.iloc[0]` asks for the first row by position.

## Exercise 7: Missing values

```python
penguins = pd.read_csv("../data/penguins.csv")

print(penguins.isna().sum())
print(penguins["sex"].isna().mean())
print(penguins["bill_length_mm"].mean())
print(penguins["bill_length_mm"].mean(skipna=False))
print(len(penguins.dropna()))
penguins["sex"].value_counts(dropna=False)
```

## Exercise 8: Translate from R

```python
flights = pd.read_csv("../data/flights.csv")

print(flights.shape)
print(flights["dep_delay"].isna().sum())
print(flights["dep_delay"].mean())
flights["origin"].value_counts()
```

No `na.rm` equivalent is needed: `.mean()` skips missing values by default.
